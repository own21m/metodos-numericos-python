```python
import matplotlib.pyplot as plt
import numpy as np
def f(y, t):
  return y * math.sin(t)**3

def euler(f, t0, y0, t_f, h):
  """
  Implementación del método de Euler para resolver ecuaciones diferenciales ordinarias.

  Args:
    f: La función a integrar.
    t0: El punto inicial.
    y0: El valor inicial de la solución.
    t_f: El punto final.
    h: El paso de integración.

  Returns:
    Una lista con los valores de la solución en los puntos de integración.
  """

  y = [y0]

  for t in np.arange(t0, t_f, h):
    y.append(y[-1] + h * f(y[-1], t))

  return y

def heun(f, t0, y0, t_f, h):
  """
  Implementación del método de Heun para resolver ecuaciones diferenciales ordinarias.

  Args:
    f: La función a integrar.
    t0: El punto inicial.
    y0: El valor inicial de la solución.
    t_f: El punto final.
    h: El paso de integración.

  Returns:
    Una lista con los valores de la solución en los puntos de integración.
  """

  y = [y0]

  for t in np.arange(t0, t_f, h):
    k1 = h * f(y[-1], t)
    k2 = h * f(y[-1] + k1, t + h)
    y.append(y[-1] + (k1 + k2) / 2)

  return y

def fanalitic(t):
  return math.e**(-math.cos(t)+ ((math.cos(t))**3)/ 3)

y_euler = euler(f, 0, 1, 20, 0.1)
y_heun = heun(f, 0, 1, 20, 0.1)
y_analitic = [fanalitic(t) for t in np.arange(0, 20, 0.1)]

print("y_euler: ", y_euler)
print("(y_heun: ", y_heun)
print("y_analitic: ", y_analitic)

plt.plot(y_euler, label="Euler")
plt.plot(y_heun, label="Heun")
plt.plot(y_analitic, label="Analítica")
plt.legend()
plt.show()
```
