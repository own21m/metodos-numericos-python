```python
import matplotlib.pyplot as plt
import numpy
c = 0
p, mx, my = [], [], []
while True:
    c += 1
    px = input("Ingrese x de la coordenada: ")
    if px == "": break
    py = input("Ingrese y de la coordenada: ")
    if py == "": break
    my.append(float(py))
    pa = [float(px), float(py)]
    p.append(pa)
for i in range(len(p)):
    n = 1
    ma = []
    for j in range(len(p)):
        if j == 0: ma.append(1)
        elif j != 0:
            ma.append((p[i][0])**n)
            n += 1
    mx.append(ma)
print("Matriz de coeficientes: ", mx)
print("Matriz de resultados: ", my)
r = numpy.linalg.solve(mx, my)
print("Coeficientes resultantes: ", r)
def polinomy(x):
    n = 0
    for k in range(len(r)):
        if k ==0: ply = r[k]*x**n
        else: ply += r[k]*x**n
        n += 1
    return ply
equis = []
for g in range(1,8):
    equis.append(polinomy(g))
plt.plot(equis, label="x", linestyle="dotted", linewidth=2)
plt.legend()
plt.show()
```
