```python
import numpy as np
from scipy.linalg import solve

# Definir las constantes y los valores iniciales
D = 1.0
U = 0.0
k = 0.0
xa = 0.0
xb = 1.0
Ya = 0.0
Yb = 1.0
N = 10  # Número de puntos de discretización

# Calcula el espaciado entre los puntos
h = (xb - xa) / N

# Crear una matriz de coeficientes y un vector de términos independientes
A = np.zeros((N - 1, N - 1))
b = np.zeros(N - 1)

# Rellenar la matriz A y el vector b
for i in range(1, N):
    xi = xa + i * h
    A[i - 1, i - 1] = -2 * D / (h**2) - k
    if i < N - 1:
        A[i - 1, i] = D / (h**2) + U / (2 * h)
    if i > 1:
        A[i - 1, i - 2] = D / (h**2) - U / (2 * h)
    b[i - 1] = -k * xi

# Aplicar las condiciones de contorno
b[0] -= (D / (h**2) - U / (2 * h)) * Ya
b[N - 2] -= (D / (h**2) + U / (2 * h)) * Yb

# Resolver el sistema de ecuaciones lineales
y_interior = solve(A, b)

# Agregar los valores de las condiciones iniciales en los extremos
y = np.insert(y_interior, 0, Ya)
y = np.append(y, Yb)

# Imprimir la solución numérica
for i in range(N + 1):
    xi = xa + i * h
    print(f'y({xi}) = {y[i]}')
```
